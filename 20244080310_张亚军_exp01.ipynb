```python
#1 score_level
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须在0‑100之间")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

print(score_level(95))
print(score_level(82))
print(score_level(70))
print(score_level(50))

#2 for while求1‑100和
sum_for = 0
for i in range(1,101):
    sum_for +=i
print("for循环求和", sum_for)

sum_while = 0
n=1
while n<=100:
    sum_while +=n
    n +=1
print("while循环求和", sum_while)

#3 迭代、递归阶乘
def fact_iter(n):
    res =1
    for i in range(2,n+1):
        res *=i
    return res

def fact_recur(n):
    if n ==0:
        return 1
    return n * fact_recur(n-1)

for num in [0,1,5,10]:
    print(f"n={num}, iter={fact_iter(num)}, recur={fact_recur(num)}")

#4 summary 返回最小最大平均
def summary(values):
    minv = min(values)
    maxv = max(values)
    avg = sum(values)/len(values)
    return (minv, maxv, avg)

vals = [10,20,30,40]
mi,ma,av = summary(vals)
print("summary解包：", mi,ma,av)

#5 sum_all可变参数
def sum_all(*args):
    return sum(args)

print(sum_all())
print(sum_all(5))
print(sum_all(1,2,3,4))
```

```python
import os
import pickle

#1 工作目录，创建文件夹
cwd = os.getcwd()
print("当前工作目录", cwd)
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)

# 准备原始文本内容
txt_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

stu_path = os.path.join(data_dir, "students.txt")
#2 写入文件
with open(stu_path, "w", encoding="utf‑8") as f:
    f.write(txt_content)

#3 解析函数
def parse_student(line):
    line = line.strip()
    if not line:
        return None
    parts = line.split(",")
    if len(parts)!=2:
        raise ValueError("字段数量不对")
    name, score_str = parts
    name = name.strip()
    if len(name)==0:
        raise ValueError("姓名为空")
    try:
        sc = int(score_str.strip())
    except ValueError:
        raise ValueError("成绩不是整数")
    if not (0<= sc <=100):
        raise ValueError(f"成绩{sc}超出0‑100")
    return {"name":name, "score":sc}

#4 逐行处理捕获异常
valid_records = []
error_log = []
with open(stu_path, "r", encoding="utf‑8") as f:
    lines = f.readlines()
for lineno, line in enumerate(lines, start=1):
    if lineno ==1:
        continue #跳过表头
    try:
        res = parse_student(line)
        if res is not None:
            valid_records.append(res)
    except ValueError as err:
        error_log.append({"line":lineno, "err":str(err)})

print("有效记录：", valid_records)
print("错误记录：", error_log)

#5 report.txt w模式覆盖，a模式追加
valid_num = len(valid_records)
invalid_num = len(error_log)
if valid_num>0:
    avg_sc = sum(r["score"] for r in valid_records)/valid_num
    pass_c = sum(1 for r in valid_records if r["score"]>=60)
    pass_r = pass_c / valid_num
else:
    avg_sc =0
    pass_r=0

report_path = os.path.join(data_dir, "report.txt")
# w模式覆盖旧文件
with open(report_path, "w", encoding="utf‑8") as f:
    f.write(f"有效记录数:{valid_num}\n")
    f.write(f"无效记录数:{invalid_num}\n")
    f.write(f"平均分:{avg_sc:.2f}\n")
    f.write(f"及格率:{pass_r:.1%}\n")

# a追加模式演示
with open(report_path, "a", encoding="utf‑8") as f:
    f.write("---使用a模式追加一行---\n")

#6 pickle序列化
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path,"wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path,"rb") as f:
    loaded_data = pickle.load(f)
print("pickle读取校验：", loaded_data == valid_records)

#8 捕获FileNotFoundError
try:
    with open("not_exist.txt","r") as f:
        pass
except FileNotFoundError as e:
    print("捕获文件不存在异常：", e)
```

```python
import copy

#1 Student类
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >=60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score}, passed={self.is_passed()})"

#2 根据实验4有效记录创建对象
for rec in valid_records:
    s = Student(rec["name"], rec["score"])
    print(s)

#3 GraduateStudent子类继承
class GraduateStudent(Student):
    def __init__(self,name,score, research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.research_topic})"

g = GraduateStudent("Li",88,"AI")
print(g)

#4 直接赋值、copy.copy、copy.deepcopy对比
original = {"name":"Alice", "scores":[85,90]}

#直接赋值
a1 = original
a1["scores"][0]=999
print("直接赋值修改内部列表：", original)
original["scores"][0]=85 #复原

#浅拷贝
a2 = copy.copy(original)
a2["scores"][0]=999
print("copy.copy浅拷贝：", original)
original["scores"][0]=85

#深拷贝
a3 = copy.deepcopy(original)
a3["scores"][0]=999
print("copy.deepcopy深拷贝：", original)
```
